## Business Problem: Production Optimization

A factory manufactures two products (A and B).  
Each product requires machine hours and generates profit.

### Objective:
Maximize total profit while respecting machine and production constraints.


```python
from pulp import LpMaximize, LpProblem, LpVariable, lpSum
```

## Step 1: Define Optimization Model
We define a maximization problem since we want to maximize profit.


```python
model = LpProblem("Production_Optimization", LpMaximize)
```

## Step 2: Decision Variables
Let:
- x = units of Product A
- y = units of Product B


```python
x = LpVariable("Product_A", lowBound=0, upBound=15)
y = LpVariable("Product_B", lowBound=0, upBound=30)
```

## Step 3: Objective Function
Maximize Profit = 40x + 30y


```python
model += 40*x + 30*y
```

## Step 4: Constraints
- Machine hours constraint
- Maximum production limits


```python
model += 2*x + 1*y <= 40   # Machine hours constraint
```

## Step 5: Solve the Model


```python
model.solve()
```

## Step 6: Results


```python
print("Status:", model.status)
print("Optimal Production of Product A:", x.value())
print("Optimal Production of Product B:", y.value())
print("Maximum Profit:", model.objective.value())
```

## Business Insights

- The optimal production strategy maximizes profit while respecting constraints.
- Linear Programming helps businesses make data-driven decisions.
- This model can be extended to include labor cost, raw material limits, and demand forecasting.
