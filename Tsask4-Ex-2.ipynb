**Technique:** Linear Programming  
**Library:** PuLP


## Business Problem: Transportation Optimization

The objective is to minimize transportation cost while satisfying
warehouse supply and store demand constraints.


```python
from pulp import LpMinimize, LpProblem, LpVariable, lpSum
```

## Step 1: Create Minimization Model


```python
model = LpProblem("Transportation_Optimization", LpMinimize)
```

## Step 2: Decision Variables
x[i][j] = units shipped from Warehouse i to Store j


```python
warehouses = ["W1", "W2"]
stores = ["S1", "S2", "S3"]

x = LpVariable.dicts("Ship",
                     (warehouses, stores),
                     lowBound=0)
```

## Step 3: Transportation Cost Matrix


```python
cost = {
    ("W1", "S1"): 4, ("W1", "S2"): 6, ("W1", "S3"): 9,
    ("W2", "S1"): 5, ("W2", "S2"): 4, ("W2", "S3"): 7
}
```

## Step 4: Objective Function
Minimize total transportation cost


```python
model += lpSum(cost[(w, s)] * x[w][s] for w in warehouses for s in stores)
```

## Step 5: Constraints
- Supply constraints
- Demand constraints


```python
# Supply constraints
supply = {"W1": 50, "W2": 60}
for w in warehouses:
    model += lpSum(x[w][s] for s in stores) <= supply[w]

# Demand constraints
demand = {"S1": 30, "S2": 40, "S3": 40}
for s in stores:
    model += lpSum(x[w][s] for w in warehouses) == demand[s]
```

## Step 6: Solve the Model


```python
model.solve()
```

## Step 7: Optimal Shipping Plan


```python
print("Minimum Transportation Cost =", model.objective.value())

for w in warehouses:
    for s in stores:
        if x[w][s].value() > 0:
            print(f"Ship {x[w][s].value()} units from {w} to {s}")
```

## Business Insights

- Optimal allocation reduces logistics cost significantly.
- Linear programming ensures all store demands are met.
- Model can be extended for fuel costs, delivery time, or vehicle limits.
